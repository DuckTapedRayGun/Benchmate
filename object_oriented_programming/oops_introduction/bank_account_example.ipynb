# Without OOPs

```python
# Import Functions from the bank_account_functions module
from bank_account_functions import withdraw, deposit, add_transaction
```

```python
# Create a variable for the balance of a person's bank account
ankit_balance = 1000

# Create a variable to keep track of ankit's transactions
ankit_transactions = []

# Add a transaction to Ankit's account
ankit_transactions = add_transaction(ankit_balance, ankit_transactions)

print("Ankit's balance:", ankit_balance)
print("Ankit's transactions:", ankit_transactions)
```

```python
# Withdraw money from Ankit's account
ankit_balance = withdraw(ankit_balance, 100)
ankit_transactions = add_transaction(ankit_balance, ankit_transactions)

print("Ankit's balance:", ankit_balance)
print("Ankit's transactions:", ankit_transactions)
```

```python
# Deposit money into Ankit's account
ankit_balance = deposit(ankit_balance, 200)
ankit_transactions = add_transaction(ankit_balance, ankit_transactions)

print("Ankit's balance:", ankit_balance)
print("Ankit's transactions:", ankit_transactions)
```
